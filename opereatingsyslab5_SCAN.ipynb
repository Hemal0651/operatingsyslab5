```python
req_seq = [int(x) for x in input("Enter numbers separated by spaces: ").split()]
head=int(input("Any head number: "))
total_seek=0
sequence=[]
remaining_requests=req_seq.copy()
while remaining_requests:
  i=min(remaining_requests, key=lambda r:abs(r-head))
  total_seek+=abs(i-head)
  head=i
  sequence.append(i)
  remaining_requests.remove(i)
print(total_seek)
print(sequence)
```
