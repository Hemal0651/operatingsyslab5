```python
head = int(input("Any head number: "))
reqs = list(map(int, input("Requests: ").split()))
direction = input("Direction (left or right): ").lower()

left = sorted([x for x in reqs if x < head], reverse=True)
right = sorted([x for x in reqs if x >= head])

if direction == "left":
    if 0 not in left:
        left.append(0)
    path = left + right

elif direction == "right":
    if 199 not in right:
        right.append(99)
    path = right + left

else:
    print("Invalid direction. Please enter 'left' or 'right'.")
    exit()

total_movement = 0
current = head

for track in path:
    total_movement += abs(current - track)
    current = track

print("Order:", [head] + path)
print("Total head movement:", total_movement)
```
